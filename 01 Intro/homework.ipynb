```python
#Get the underlying NumPy array. Let's call it X.


import numpy as np # import numpy

X = np.array(df[['RAM','Storage','Screen']] )
```

```python
X.ndim
```

```python
# Compute matrix-matrix multiplication between the transpose of X and X.
# To get the transpose, use X.T. Let's call the result XTX.

XTX  = np.dot(X.T,X)
```

```python
def is_square_numpy_matrix(matrix):
    """Take a matrix as input ,checks if it is square and returns a True or False value"""
    if not isinstance(matrix, np.ndarray):
        raise TypeError("Input must be a NumPy array.")
    
    if matrix.ndim != 2:  # Ensure it's a 2D matrix
        return False
        
    rows, cols = matrix.shape
    return rows == cols
```

```python
is_square_numpy_matrix(XTX)
```

```python
#Compute the inverse of XTX.

XTX_inv = np.linalg.inv(XTX)
```

```python
#Create an array y with values [1100, 1300, 800, 900, 1000, 1100].

y = np.array([1100, 1300, 800, 900, 1000, 1100])
y
```
